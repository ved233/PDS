# Practical 3
### Aim
To implement programs using if, if-else, elif, while loop, for loop, nested loops and loop control statements.

### 1. if Statement

```python
age = 21

if age >= 18:
    print("You can vote")
```

### 2. if-else Statement

```python
number = 41

if number % 2 == 0:
    print("Even")
else:
    print("Odd")
```

### 3. elif Statement

```python
marks = 85

if marks >= 90:
    print("A")
elif marks >= 75:
    print("B")
elif marks >= 50:
    print("C")
else:
    print("Fail")
```

### 4. while Loop

```python
i = 1

while i <= 10:
    print(i)
    i = i + 1
```

### 5. for Loop

```python
for i in range(1, 11):
    print(i)
```

### 6. Sum using for Loop

```python
total = 0

for i in range(1, 11):
    total = total + i

print(total)
```

### 7. Nested Loop

```python
for i in range(1, 4):
    for j in range(1, 4):
        print(i, j)
```

### 8. Nested Loop Pattern

```python
for i in range(1, 6):
    for j in range(i):
        print("*", end=" ")
    print()
```

### 9. break Statement

```python
for i in range(1, 11):
    if i == 6:
        break
    print(i)
```

### 10. continue Statement

```python
for i in range(1, 11):
    if i == 5:
        continue
    print(i)
```

### 11. pass Statement

```python
for i in range(1, 6):
    if i == 3:
        pass
    print(i)
```

### 12. Check Positive, Negative or Zero

```python
number = int(input("Enter a number: "))

if number > 0:
    print("Positive")
elif number < 0:
    print("Negative")
else:
    print("Zero")
```
