### Aim
To demonstrate exception handling using try, except, else, finally blocks and raise statement with custom exceptions.

### 1. try and except

```python
try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    print(a / b)
except ZeroDivisionError:
    print("Cannot divide by zero")
```

### 2. Handling ValueError

```python
try:
    age = int(input("Enter age: "))
    print("Age:", age)
except ValueError:
    print("Enter a valid number")
```

### 3. try, except and else

```python
try:
    a = int(input("Enter number: "))
    b = int(input("Enter number: "))
    result = a + b
except ValueError:
    print("Invalid input")
else:
    print("Sum:", result)
```

### 4. try, except, else and finally

```python
try:
    a = int(input("Enter number: "))
    b = int(input("Enter number: "))
    print(a / b)
except ZeroDivisionError:
    print("Cannot divide by zero")
else:
    print("Division completed")
finally:
    print("Program finished")
```

### 5. Multiple except Blocks

```python
try:
    a = int(input("Enter number: "))
    b = int(input("Enter number: "))
    print(a / b)
except ValueError:
    print("Enter numbers only")
except ZeroDivisionError:
    print("Cannot divide by zero")
```

### 6. raise Statement

```python
age = 15

try:
    if age < 18:
        raise ValueError("Age must be 18 or above")
    print("Eligible")
except ValueError as e:
    print(e)
```

### 7. Custom Exception

```python
class AgeError(Exception):
    pass

age = 16

try:
    if age < 18:
        raise AgeError("Age is less than 18")
    print("Eligible")
except AgeError as e:
    print(e)
```

### 8. Custom Exception with User Input

```python
class MarksError(Exception):
    pass

marks = int(input("Enter marks: "))

try:
    if marks < 0 or marks > 100:
        raise MarksError("Marks should be between 0 and 100")
    print("Valid marks")
except MarksError as e:
    print(e)
```
